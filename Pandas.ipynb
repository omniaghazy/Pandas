# Missing Values


**Detect**

```python
df.isna().sum()
```

**Drop**

```python
df.dropna()       #drop row that has at least one cell null
# لكن كده مش هيعحصل تغيير
```

```python
df = df.dropna()
```

لو عايز امسح الروو اللي كله فاضي

```python
# df.dropna(
#     axis = 0      for row (default )
#          = 1       for col      ,

#     how = any       nill  اي صف فيه علي الاقل مره
#         = all          null    لازم كل الصف يكون  ,

#     thresh = n                             من القيم فعلا موجوده   n   الصف علي الاقل فيه عدد
#     ,

#     subset=[col 2 , col 3]                            بنحدد اي اعمده عايزنها

#     ,


#     inplace = True                              يعني التغير بتاع الدروب يسمع في الفايل الاصلي
#              =False                                    التعديل ميسمعش في الاصلي
# )
```

```python
df.dropna(subset =["order_source" ,"order_date"]).head()
```

**fill missing value**

```python
df.fillna(value="not found" ,inplace = True)
```

```python
df.fillna(value={
    "order_source" : "Web",
    "sales" : df.sales.median()
},
          inplace =True
          )
```
